```python
from datasets import load_dataset
import pandas as pd
from tqdm import tqdm

LIMIT = 2_100_000

dataset = load_dataset(
    "ai4bharat/IndicCorpV2",
    split="hin_Deva",
    streaming=True
)

texts = []

for i, sample in enumerate(tqdm(dataset)):
    # Depending on dataset format
    if "text" in sample:
        texts.append(sample["text"])
    else:
        texts.append(list(sample.values())[0])

    if i + 1 == LIMIT:
        break

df = pd.DataFrame({"text": texts})

print(df.head())
print(len(df))
```

```python
df.drop_duplicates(subset=["text"], inplace=True)
df.dropna(subset=['text'], inplace=True)

updated_df = df[df['text'] != '']
updated_df.reset_index(drop=True, inplace=True)
updated_df
```

```python
updated_df.to_parquet(
    "hindi_texts_raw.parquet",
    engine="pyarrow",
    compression="snappy",
    index=False
)
```

```python
import re
```

```python
def sentence_tokenizer(text):
  pattern = r'[^।!?॥]+[।!?॥]*'

  sentences = re.findall(pattern, text)
  return [
    sentence.strip() for sentence in sentences
  ]
```

```python
sentence_tokenizer(df.iloc[1, 0])
```

```python
df['text'].shape
```

```python
tot_sentences = []
for txt in df.loc[:, 'text']:
  sentences = sentence_tokenizer(txt)
  tot_sentences.extend(sentences)
```

```python
len(tot_sentences)
```

```python
TOKEN_PATTERN = re.compile(
    r"""
    https?://\S+                     # URLs
    | www\.\S+
    | [A-Za-z0-9._%+-]+@[A-Za-z0-9.-]+\.[A-Za-z]{2,}   # Email
    | \d{1,2}[/-]\d{1,2}[/-]\d{2,4}   # Dates
    | \d+\.\d+                       # Decimal
    | \d+                            # Integer
    | [.,!?;:"'(){}\[\]<>/\-—–।॥…%&@#$^*_+=|\\`~।]                # Punctuation
    | [\u0900-\u0963\u0966-\u097F]+               # Hindi words
    """,
    re.VERBOSE
)
  
```

```python
def word_tokenizer(sentence):
  return re.findall(TOKEN_PATTERN, sentence)
```

```python
word_tokenizer("यह एक rajdeeppa53@gmail.com वाक्य है। यह दूसरा 12/09/2025 वाक्य है।")
```

```python
data_df = pd.DataFrame({"sentence": tot_sentences})
data_df.head()
```

```python
data_df['words'] = data_df['sentence'].apply(word_tokenizer)
data_df['words_count'] = data_df['words'].apply(len)
data_df.head()
```

```python
data_df.to_parquet(
  "hindi_texts_tokenized.parquet",
  engine="pyarrow",
  compression="snappy",
  index=False
)
```

```python
df=pd.read_parquet(
  "hindi_texts_tokenized.parquet"
)

df.head()
```

```python
total_sentences=df.shape[0]
total_sentences
```

```python
total_words=df['words_count'].sum()
total_words
```

```python
total_characters = 0
for i in range(total_sentences):
  total_characters += len(df['sentence'][i])

total_characters
```

```python
avg_sentence_length = total_words / total_sentences if total_sentences > 0 else 0
avg_sentence_length
```

```python
avg_word_length = total_characters / total_words if total_words > 0 else 0
avg_word_length
```

```python
all_tokens = []
df['words'].apply(lambda x: all_tokens.extend(x))
total_number_of_tokens = len(all_tokens)

all_unique_tokens = list(set(all_tokens))
total_number_of_unique_tokens = len(all_unique_tokens)
total_number_of_tokens, total_number_of_unique_tokens
```

```python
TTR = total_number_of_unique_tokens / total_number_of_tokens if total_number_of_tokens > 0 else 0
TTR
```

```python
print(f"i.   Total number of sentences: {total_sentences:,}")
print(f"ii.  Total number of words: {total_words:,}")
print(f"iii. Total number of characters: {total_characters:,}")
print(f"iv.  Average Sentence Length: {avg_sentence_length:.2f} words/sentence")
print(f"v.   Average word length: {avg_word_length:.2f} characters/word")
print(f"vi.  Type/Token Ratio (TTR): {TTR:.6f} ({total_number_of_unique_tokens:,} unique / {total_number_of_tokens:,} total)")
```
